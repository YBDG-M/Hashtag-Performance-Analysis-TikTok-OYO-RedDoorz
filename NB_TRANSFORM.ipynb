```python
import pandas as pd

# Baca file CSV
df = pd.read_csv("D:\DATA ANALYST-DATA SCIENTIST\MY PROJECT 2025\MARKETING FILE\OYO VS REDDOORZ\Datest_Reddoorz.csv", encoding="utf-8")

# Cari kolom hashtags yang berformat hashtags/0/name, hashtags/1/name, dst
hashtag_cols = [col for col in df.columns if col.startswith("hashtags/")]

# Gabungkan semua hashtags jadi satu kolom
df["hashtags"] = df[hashtag_cols].apply(
    lambda row: " ".join(["#" + str(tag) for tag in row if pd.notna(tag)]),
    axis=1
)

# Pilih kolom yang dibutuhkan + hashtags gabungan
df_out = df[[
    "text",
    "diggCount",
    "shareCount",
    "playCount",
    "commentCount",
    "authorMeta/name",
    "webVideoUrl",
    "createTimeISO",
    "collectCount",
    "hashtags"
]]

# Simpan ke CSV baru
df_out.to_csv("Data_clean_reddoorz.csv", index=False, encoding="utf-8-sig")

print("✅ File output.csv sudah jadi, bisa dibuka di Excel.")
```
